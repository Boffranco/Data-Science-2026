```python
import pandas as pd
import glob #encontrar la ruta
import os #interacturar con el so para minupular archivos
```

```python
ruta_csv = "data/raw/*.csv"
```

```python
archivos_encontrados = glob.glob(ruta_csv)
```

```python
lista_dfs = []
for archivo in archivos_encontrados:
    try:
        df_temp = pd.read_csv(archivo, on_bad_lines='skip', engine='python', encoding='latin1')
        df_temp["temporada_origen"] = os.path.basename(archivo)
        lista_dfs.append(df_temp)
    except Exception as e:
        # Si algún archivo falla por completo, te avisará cuál es sin detener el resto
        print(f"No se pudo procesar el archivo {archivo}. Error: {e}")

print(f"Se cargaron {len(lista_dfs)} archivos exitosamente.")
```

```python
df_maestro = pd.concat(lista_dfs, ignore_index=True)
```

```python
print(f"Archivos procesados: {len(archivos_encontrados)}")
print(f"Tamaño total del DataFrame: {df_maestro.shape}")
display(df_maestro.head())
```
