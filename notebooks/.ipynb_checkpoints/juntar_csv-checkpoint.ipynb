```python
import pandas as pd
import glob #encontrar la ruta
import os #interacturar con el so para minupular archivos
```

```python
ruta_csv = "../data/raw/*.csv"
```

```python
archivos_encontrados = glob.glob(ruta_csv)
```

```python
lista_dfs = []
for archivo in archivos_encontrados:
    df_temp = pd.read_csv(archivo)
    df_temp["temporada_origen"] = os.path.basename(archivo)
    lista_dfs.append(df_temp)
```

```python
df_maestro = pd.concat(lista_dfs, ignore_index=True)
```
